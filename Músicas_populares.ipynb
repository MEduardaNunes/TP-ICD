### Descrição dos atributos dos bancos de dados
Uma breve descrição dos **ATRIBUTOS** compartilhados pelos conjuntos de dados:

* **track_name** **(nome_da_faixa) :** Nome da faixa musical

* **artist_name** **(nome_do_artista) :** Nome do(s) artista(s) creditados na faixa

* **track_pop:** **(popularidade_da_faixa) :** Popularidade da faixa, quanto maior o número, maior a popularidade da faixa com base no número de ouvintes e reproduções (medida de 0 a 100).

* **danceability** **(dançabilidade) :** A dançabilidade descreve o quão adequada uma faixa é para dançar com base em uma combinação de elementos musicais, quanto maior o número, mais dançante é a faixa (medida de 0 a 1).

* **energy** **(energia) :** A Energia é representa uma medida perceptiva de intensidade e atividade. Normalmente, faixas energéticas parecem rápidas, altas e barulhentas, quanto maior o valor, mais energética é a faixa (medida de 0 a 1).

* **loudness** **(volume) :** O volume total de uma faixa em decibéis (dB). Os valores normalmente variam entre -60 e 0 dB, conforme a escala para sistemas digitais (medida de 0 a 1).

* **key** **(chave) :** A tonalidade da faixa indica qual a escala musical predominante na faixa, dependendo da chave utilizada, uma música pode ter uma sensação totalmente diferente (apresentada utilizando representação numérica).

* **speechiness** **(fala) :** Detecta a presença de palavras faladas em uma faixa, quanto maior o número, maior a probabilidade da faixa conter palavras faladas, como em podcasts ou audiolivros (medida de 0 a 1).

* **acousticness** **(acústica) :**  Determina o quão acústica é uma faixa, quanto menor mais eletrônica ela é, e quanto maior, maior a presença de instrumentos acústicos (medida de 0 a 1).

* **instrumentalness** **(instrumentalidade) :** Analisa a presença de voz em uma faixa, quanto maior, menor é a presença de voz na faixa e mais provável de ela ser uma faixa instrumental (medida de 0 a 1).

* **liveness** **(vivacidade) :** Mede a probabilidade de uma faixa musical ter sido gravada ao vivo, na presença de uma audiência, quanto maior, maior essa probabilidade (medida de 0 a 1).

* **tempo** **(tempo) :** O andamento estimado geral de uma faixa em batidas por minuto (BPM), quanto maior, maior a velocidade e o ritmo da faixa.

* **duration_ms** **(duração_ms) :** A duração da faixa em milissegundos.

```python
# Importação banco de dados spotify de 1921 a 2020 e de 2021
path_spt_1921_2020 = kagglehub.dataset_download("yamaerenay/spotify-dataset-19212020-600k-tracks")
path_spt_2021 = kagglehub.dataset_download("equinxx/spotify-top-50-songs-in-2021")

# Manipulação dos Caminhos
csv_path_spt_1921_2020_2 = os.path.join(path_spt_1921_2020, "tracks.csv")
csv_path_spt_2021 = os.path.join(path_spt_2021, "spotify_top50_2021.csv")

# Leitura dos dados
df_spotify_1 = pd.read_csv(csv_path_spt_1921_2020_2)
df_spotify_2 = pd.read_csv(csv_path_spt_2021)

# Renomeia colunas para manter consistência
df_spotify_1 = df_spotify_1.rename(columns={
    'name': 'track_name',
    'artists': 'artist_name',
    'popularity': 'track_pop'
})

df_spotify_2 = df_spotify_2.rename(columns={
    'popularity': 'track_pop'
})

# Concatena tudo em uma tabela só
df_spotify = pd.concat([df_spotify_1, df_spotify_2], ignore_index=True)

# Remove colunas desnecessárias para a análise
df_spotify = df_spotify.drop(columns=['id_artists', 'explicit', 'track_id','mode','valence','time_signature','id'], errors='ignore')
```

Além disso, para melhorar a legibilidade do código, algumas funções que serão utilizadas foram definidas a seguir.

```python
# Função para calcular intervalo de confiança de 95%
def IC(dados):
    n = len(dados)
    media = np.mean(dados)
    desvio_padrao = np.std(dados, ddof=1)
    erro_padrao = desvio_padrao / np.sqrt(n)
    ic_inferior = media - 1.96 * erro_padrao
    ic_superior = media + 1.96 * erro_padrao
    return ic_inferior, ic_superior

# Função para fazer bootstrap de média, calculando intervalo de confinaça de 95%
def bootstrap(data, n_sub, size, col):
    values = np.zeros(n_sub)
    for i in range(n_sub):
        sample = data.sample(size, replace=True)
        values[i] = sample[col].mean()

    # Gerando valores inferior e superior para um nível de confiança de 95%
    LI = np.percentile(values, 2.5)
    LS = np.percentile(values, 97.5)
    return values, LI, LS

# Função para gerar análises com bootstrap e plots de determinados dados
def analise_bootstrap(df, colunas, nome_colunas):
  valores_bs_geral = []
  for coluna in colunas:
    print(nome_colunas[colunas.index(coluna)])

    # Duração média de uma música popular
    media = df[coluna].mean()
    desvio_padrao = df[coluna].std()

    print(f"{nome_colunas[colunas.index(coluna)]} média de músicas populares:", np.round(media, 4))
    print("Desvio padrão de músicas populares:", np.round(desvio_padrao, 2))

    # Intervalo de confiança de 95% para a média
    LI_original, LS_original = IC(df[coluna])
    print(f"Intervalo de confiança de 95% para a média de {nome_colunas[colunas.index(coluna)].lower()} de músicas populares: [{np.round(LI_original, 4)}, {np.round(LS_original, 4)}]")

    # Será que tal média é devido ao acaso?
    # Hipótese nula: a média é explicada ao acaso
    # Hipótese alternativa: a média não é explicada ao acaso
    # Para isso, como temos apenas uma amostra, podemos gerar várias subamostras e calcular a média das médias
    n_sub = 5000
    size = len(df)
    valores_bs, LI_bs, LS_bs = bootstrap(df, n_sub, size, coluna)
    valores_bs_geral.append(valores_bs)
    print('Intervalo de confiança do BootStrap: [{}, {}]'.format(LI_bs.round(4), LS_bs.round(4)))

    if (LI_bs <= media <= LS_bs):
      print('Não rejeitamos a hipótese nula')
    else:
      print('Rejeitamos a hipótese nula')

    # Plotando para melhor visualização
    cores = sns.color_palette("Set2", len(colunas))
    fig, axs = plt.subplots(1, 2, figsize=(12, 4))
    axs[0].hist(df[coluna], bins=20, edgecolor='black', color=cores[colunas.index(coluna)])
    axs[0].set_xlabel(f'{nome_colunas[colunas.index(coluna)]} média')
    axs[0].set_ylabel('Frequência')
    axs[0].set_title('Dados originais')

    axs[1].hist(valores_bs, bins=20, edgecolor='black', color=cores[colunas.index(coluna)])
    axs[1].set_xlabel(f'{nome_colunas[colunas.index(coluna)]} média')
    axs[1].set_ylabel('Frequência')
    axs[1].set_title('Dados gerados pelo bootstrap')
    plt.show()
    print('─' * 150)
  return valores_bs_geral

# Função de gerar o bootstrap de diferença das médias de dois dataframes e plotar
def analise_bootstrap_diferenca(df1, df2, colunas, nome_colunas):
  cores = sns.color_palette("Set2", len(colunas))
  for i, (nome_coluna, nome_atributo) in enumerate(zip(nome_colunas, colunas)):
    print(nome_coluna)

    # Bootstrap da média
    n_sub = 5000
    diferencas_medias = np.zeros(n_sub)

    for j in range(n_sub):
      # Amostras aleatórias
      sample_df1 = df1.sample(len(df1), replace=True)
      sample_df2 = df2.sample(len(df2), replace=True)

      # Médias das amostras
      media_df1 = np.mean(sample_df1[nome_atributo])
      media_df2 = np.mean(sample_df2[nome_atributo])

      #Calculando e armazenando diferença da média
      diferencas_medias[j] = media_df1 - media_df2

    # Intervalo de confiança de 95%
    LI_bs, LS_bs = np.percentile(diferencas_medias, [2.5, 97.5])
    print(f'Intervalo de confiança do Bootstrap: [{LI_bs:.4f}, {LS_bs:.4f}]')
    if (LI_bs <= 0 <= LS_bs):
      print('Não rejeitamos a hipótese nula')
    else:
      print('Rejeitamos a hipótese nula')

    # Plotando para melhor visualização
    plt.figure(figsize=(12, 4))
    plt.hist(diferencas_medias, bins=30, edgecolor='black', color=cores[i])
    plt.axvline(0, color='red', linestyle='--')
    plt.xlabel(f'Diferença da média do primeiro banco de dados - segundo banco de dados ({nome_atributo})')
    plt.ylabel('Frequência')
    plt.title(f'Distribuição Bootstrap da diferença de médias em {nome_coluna}')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    print('─' * 150)

# Função de derivada
def gradiente_mse(theta, X_r, y_r):
    erro = X_r.dot(theta) - y_r
    grad = 2 * X_r.T @ erro
    return grad.flatten()

# Função de gradiente descendente estocástico
def sgd(theta, d_fun, X, y, lambda_, tol, max_iter=10000):
    theta = np.array(theta).reshape(-1)
    erro_antigo = np.inf

    for i in range(max_iter):
        r = np.random.randint(len(y))
        X_r = X.iloc[r].values.reshape(1, -1)
        y_r = np.array([y.iloc[r]])

        grad = d_fun(theta, X_r, y_r)
        theta_novo = theta - lambda_ * grad

        erro_atual = ((X.dot(theta_novo) - y) ** 2).mean()

        if np.abs(erro_atual - erro_antigo) <= tol:
            break

        theta = theta_novo
        erro_antigo = erro_atual

    return theta

# Função que divide o dataset em dados de treinamento, validação e de teste
def split_manual(X, y, train_ratio=0.6, val_ratio=0.2, seed=42):
    np.random.seed(seed)
    n = len(y)
    indices = np.random.permutation(n)
    n_train = int(train_ratio * n)
    n_val = int(val_ratio * n)
    n_test = n - n_train - n_val

    train_idx = indices[:n_train]
    val_idx = indices[n_train:n_train+n_val]
    test_idx = indices[n_train+n_val:]

    X_train = X.iloc[train_idx].reset_index(drop=True)
    y_train = y.iloc[train_idx].reset_index(drop=True)
    X_val = X.iloc[val_idx].reset_index(drop=True)
    y_val = y.iloc[val_idx].reset_index(drop=True)
    X_test = X.iloc[test_idx].reset_index(drop=True)
    y_test = y.iloc[test_idx].reset_index(drop=True)

    return X_train, y_train, X_val, y_val, X_test, y_test

# Função que calcula a métrica r2
def avaliar_modelo(X, y, theta):
    y_pred = X.dot(theta)
    r2 = 1 - ((y - y_pred) ** 2).sum() / ((y - y.mean()) ** 2).sum()
    return r2
```

Para cada atributo da lista `colunas` foram geradas algumas métricas, intervalos de confiança e duas visualizações, explicados o que são e o porquê de terem sidos utilizados.

* <u>Média</u>: consiste no valor médio de um atributo, calculado pela fórmula da média aritmética. Esse valor foi calculado para verificarmos qual é a média de um atributo e, posteriormente, verificar se esse valor estava ou não dentro do intervalo de confiança do bootstrap.
* <u>Desvio padrão</u>: é uma medida que indica a dispersão dos dados. Ele foi calculado para utilizar no cálculo do intervalo de confiança.
* <u>Intervalo de confiança de 95% dos dados reais</u>: é um intervalo, calculado a partir da média aritmética e do desvio padrão dos dados amostrais, que, em 95% das vezes, conterá o valor verdadeiro da média da população se fossem feitas várias amostras e calculados vários intervalos de confiança.
* <u>Intervalo de confiança do Bootstrap</u>: tem o mesmo significado do intervalo de confiança de 95%, mas os dados utilizados foram gerados artificialmente pelo Bootstrap.
* <u>Plot dos dados originais</u>: é o histograma de um atributo dos dados originais, utilizado para verificar o comportamento dos dados originais.
* <u>Plot dos dados do Bootstrap</u>: é o histograma de um atributo dos dados gerados pelo Bootstrap, utilizado para verificar que realmente os dados gerados seguem a distribuição de uma normal.

**REGRESSÃO LINEAR MÚLTIPLA**


---

A seguir, foi realizado uma regressão linear de múltiplos parâmetros para tentar prever o quão popular uma música é baseado nos valores de seus atributos. Para isso, foi dividido o dataset em dados de treinamento, de validação e de teste. Em seguida, o modelo foi treinado com os dados de treinamento e consequentemente testados com os dados restantes.

```python
# Selecionando dados
X_final = df_spotify_total[colunas]
y = df_spotify_total['track_pop']
```

```python
# Faz o split manual
X_train, y_train, X_val, y_val, X_test, y_test = split_manual(X_final, y, 0.6, 0.2, seed=42)

# Normaliza usando estatísticas do conjunto de treino
mean_train = X_train.mean()
std_train = X_train.std()

X_train_norm = (X_train - mean_train) / std_train
X_val_norm = (X_val - mean_train) / std_train
X_test_norm = (X_test - mean_train) / std_train

# Adiciona a coluna bias
for df in [X_train_norm, X_val_norm, X_test_norm]:
    df.insert(0, 'bias', 1)

theta = np.zeros(X_train_norm.shape[1])

np.random.seed(42)  # para reprodutibilidade no SGD
theta_final = sgd(theta, gradiente_mse, X_train_norm, y_train, 0.01, 0.0001)
```

```python
X = [X_train_norm, X_val_norm, X_test_norm]
y = [y_train, y_val, y_test]
nome_teste = ['Treino', 'Validação', 'Teste']
for X_i, y_i, nome in zip(X, y, nome_teste):
    print(nome)
    r2 = avaliar_modelo(X_i, y_i, theta_final)
    print("R2:", r2)
```

**RESPOSTA**


---

Ao analisarmos os resultados da métrica, percebemos que o modelo linear múltiplo não é um bom modelo para os dados, visto que a métrica de R² é em torno de 0.07, indicando que o modelo consegue prever apenas 7% do padrão desses dados. Para verificar se foi um erro nosso, utilizamos modelos lineares já implementados em bibliotecas, como scikit-learn, e resultados parecidos foram observados. Além disso, também tentamos modelos não lineares, como um polinomial de baixo grau e o desempenho também não foi alto.

Por isso, acreditamos que não existe uma "receita" de características de uma música para torná-la mais popular apenas com base nos atributos dispoíveis no dataset do Spotify. Isso pode ser explicado por vários motivos, como

* Atributos ausentes das músicas que impactam sua popularidade não estão disponíveis no banco de dados.
* As músicas foram analisadas em um período longo e houveram muitas mudanças nas características, o que pode ter tornado difícil encontrar um padrão nesse banco de dados
* Fatores externos, como marketing e divulgação da música, podem afetar e explicar melhor a popularidade de uma música

Apesar disso, podemos observar que algumas características tiveram impactos na popularização da música, embora seja pequeno, quando fizemos a análise exploratória dos dados. Por exemplo, as músicas do Spotify mais famosas tendem a ser mais dançáveis.

```python
for nome, dist_spotify, dist_tiktok in zip(nome_colunas, distribuicoes_spotify, distribuicoes_tiktok):
    #1º e 3º quartil
    q1_tiktok, q3_tiktok = np.percentile(dist_tiktok, [25, 75])
    q1_spotify, q3_spotify = np.percentile(dist_spotify, [25, 75])
    intersecta = not (q3_tiktok < q1_spotify or q3_spotify < q1_tiktok)

    if intersecta:
        cor_tiktok = cor_spotify = to_rgba('#4C72B0', alpha=0.7)  # azul
    else:
        cor_tiktok, cor_spotify = to_rgba('#4C72B0', alpha=0.7), to_rgba('#DD8452', alpha=0.7)  # azul e laranja

    # Plot
    fig, ax = plt.subplots()
    box = ax.boxplot(
        [dist_tiktok, dist_spotify],
        vert=False,
        patch_artist=True,
        showfliers=False,
        tick_labels=['TikTok', 'Spotify']
    )

    box['boxes'][0].set_facecolor(cor_tiktok)
    box['boxes'][1].set_facecolor(cor_spotify)

    for element in ['medians', 'whiskers', 'caps']:
        for item in box[element]:
            item.set_color('black')

    ax.grid(False)
    ax.set_title(f"Boxplot comparativo - {nome}")
    ax.set_xlabel(f"Média da {nome.lower()}")
    plt.tight_layout()
    plt.show()
```

* terem maior vivacidade.

Chegamos a essas conclusões, pois o intervalo interquartil das características citadas acima não se interceptam, visualmente identificada por cores diferentes nos plots, e, por isso, pode indicar que realmente existe diferenças nas características predominantes nas músicas dos dois aplicativos.

Podemos verificar se essa interseção realmente não existe gerando um Bootstrap para a diferença das médias entre o Spotify e o TikTok. Caso o valor zero não esteja no intervalo de confiança gerado por esse Bootstrap, então as médias têm diferença significativa.
